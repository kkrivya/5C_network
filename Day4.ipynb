```python
def clean_text(text):
    """
    Clean a single text value:
    - Strip leading/trailing spaces
    - Convert to lowercase
    """
    cleaned = text.strip().lower()
    return cleaned

raw = "  AlIce  "
result = clean_text(raw)
print(result)
```

```python
def log_event(user, action):
    print(f"{user} did {action}")

log_event("Alex", "login")
```

```python
def create_profile(username, role="Standard"):
    """
    Builds a new user profile.
    role: defaults to 'Standard' for normal sign-ups.
    """
    print(f"Setting up a {role} account for {username}...")

    # Return a structured dictionary
    return {"user": username, "role": role, "active": True}

# 1. Normal Sign-up (Uses default 99% of the time)
new_user = create_profile("Alex")

# 2. Internal Hire (Overrides the default)
admin_user = create_profile("Sam", "Admin")
```

```python
def total_revenue(amounts):
    total = 0
    for value in amounts:
        total += value
    return total

revenues = [100.0, 250.5, 90.0]
print(total_revenue(revenues))
```

```python
def tag_vip(user):
    if user["lifetime_value"] > 50000:
        user["segment"] = "VIP"

customer = {"name": "Alex", "lifetime_value": 60000}
tag_vip(customer)
print(customer)
```

```python
def min_max(values):
    smallest = min(values)
    largest = max(values)
    return smallest, largest

nums = [3, 10, 1, 8]
low, high = min_max(nums)
print(low)
print(high)
```

```python
def summarize_sales(sales):
    total = 0
    count = 0

    for amount in sales:
        total += amount
        count += 1

    average = total / count if count > 0 else 0
    return total, average

numbers = [100.0, 250.0, 150.0]
total, avg = summarize_sales(numbers)
print(f"Total: {total}, Average: {avg}")
```

```python
tax_rate = 0.05  # Global variable

def calculate_total(price):
    # Relies on data floating around outside the function
    return price + (price * tax_rate)

# Somewhere else in a massive file, this gets changed by mistake...
tax_rate = 0.99

# Now our function is broken, and we don't know why!
print(calculate_total(100))
```

```python
def calculate_total(price, tax_rate=0.05):
    # Everything the function needs is handed directly to it
    return price + (price * tax_rate)

# Safe, predictable, and isolated
print(calculate_total(100))
```

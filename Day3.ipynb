```python
rooms_to_search = ["Kitchen", "Living Room", "Bedroom", "Bathroom"]

for room in rooms_to_search:
    print(f"Searching the {room}...")

    if room == "Bedroom":
        print("Found the keys! Stopping the search.")
        break  # <--- Exits the loop immediately!

print("Search mission over.")
```

```python
rooms_home = ["Kitchen", "Living Room", "Bedroom", "Hall"]

for room in rooms_home:
  print(f"Searching the {room}...")

  if room == "Bedroom":
    print("Found the keys! No need to search further")
    break # <---exit the loop instantly

print("Mission successful")

```

```python
apples_on_belt = ["good", "rotten", "good", "rotten", "good"]
basket = []

for apple in apples_on_belt:
    if apple == "rotten":
        print("Discarding a rotten apple...")
        continue  # <--- Skips the rest of the code and grabs the next apple!

    print("Packing a good apple.")
    basket.append(apple)

print(f"Finished! Packed {len(basket)} good apples.")
```

```python
# Danger (The Skipping Bug)
tasks = ["Write code", "Fix bug", "Fix bug", "Deploy"]

# Attempting to remove all bugs
for task in tasks:
    if task == "Fix bug":
        tasks.remove(task)
        newtask = tasks
        for task in newtask:
          if task == "Fix bug":
            newtask.remove(task)
            print(newtask)


print(tasks)
```

```python
# Safe (Building a New List)
tasks = ["Write code", "Fix bug", "Fix bug", "Fix bug", "Deploy"]
clean_tasks = []

# Safely keeping only the good tasks
for task in tasks:
    if task != "Fix bug":
        clean_tasks.append(task)

print(clean_tasks)
```
